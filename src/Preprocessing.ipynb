## Import libraries

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

## Load Dataset

```python
df=pd.read_csv("../data/insurance.csv")
print(df)
```

## Label Encoding

```python
le=LabelEncoder()
df["sex"]=le.fit_transform(df["sex"])
df["smoker"]=le.fit_transform(df["smoker"])
```

Label Encoding: Used for sex and smoker because these features have only two categories, making numerical encoding simple and suitable for the model.

## One hot encoding the Region column

```python
df = pd.get_dummies(df, columns=["region"], dtype=int)

print(df)
```

One-Hot Encoding: Used for region because its categories have no natural order, so separate binary columns avoid implying any ranking between regions.

## Save Cleaned  file

```python
df.to_csv("../data/cleaned_data.csv",index=False)
```
